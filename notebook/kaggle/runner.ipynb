```python
import base64, os
from pathlib import Path
import nbformat
from nbclient import NotebookClient
source=Path('/kaggle/working/AdaZero_DCE_Complete_Evaluation.ipynb')
source.write_bytes(base64.b64decode(''))
nb=nbformat.read(source,as_version=4)
client=NotebookClient(nb,timeout=1800,kernel_name='python3',resources={'metadata':{'path':'/kaggle/working'}})
try:
    client.execute()
finally:
    nbformat.write(nb,'/kaggle/working/AdaZero_DCE_Complete_Evaluation_Executed.ipynb')
errors=[o for c in nb.cells if c.cell_type=='code' for o in c.get('outputs',[]) if o.output_type=='error']
assert not errors, errors
print('Executed notebook saved with',sum(len(c.get('outputs',[])) for c in nb.cells),'real output blocks')
from nbconvert import HTMLExporter
html,_=HTMLExporter().from_notebook_node(nb)
Path('/kaggle/working/AdaZero_DCE_Complete_Evaluation_Executed.html').write_text(html)
```
